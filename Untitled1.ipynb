```python
# Migration matrix
#
# Columns = where people are coming FROM
# Rows    = where people are going TO

A = np.array([
    [0.8, 0.1, 0.2],
    [0.1, 0.8, 0.1],
    [0.1, 0.1, 0.7]
])

print("Migration matrix A:")
print(A)

print("\nCheck: each column should add to 1")
print(A.sum(axis=0))
```
